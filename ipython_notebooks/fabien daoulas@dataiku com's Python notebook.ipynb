```python
import dataiku
from dataiku import pandasutils as pdu
import pandas as pd
```

```python
budget_joined_prepared = dataiku.Dataset("budget_joined_prepared")
budget_joined_prepared_df = budget_joined_prepared.get_dataframe()
over_budget_df = budget_joined_prepared_df[
    (budget_joined_prepared_df["isOverBudget"] == 1) | (budget_joined_prepared_df["hasTwentyPercentBudgetLeft"] == 1)
]

over_budget_df.loc[:, "overBudgetNotificationSent"] = False
over_budget_df.loc[:, "twentyPercentLeftNotificationSent"] = False

over_budget_notifications = dataiku.Dataset("over_budget_notifications")
try:
    over_budget_notifications_df = over_budget_notifications.get_dataframe()
except:
    over_budget_notifications_df = None

# keep track of notifications already sent.
if over_budget_notifications_df is not None:
    for index, row in over_budget_notifications_df.iterrows():
        project_key = row["projectKey"]
        incoming_budget_usd = over_budget_df.loc[
            over_budget_df["projectKey"] == project_key
        ]["budgetUSD"].values[0]
        if incoming_budget_usd == row["budgetUSD"]:
            print(row["overBudgetNotificationSent"])
            over_budget_df.loc[over_budget_df["projectKey"] == project_key, "overBudgetNotificationSent"] = row["overBudgetNotificationSent"]
            over_budget_df.loc[over_budget_df["projectKey"] == project_key, "twentyPercentLeftNotificationSent"] = row["twentyPercentLeftNotificationSent"]
        else:
            print("toto")
```
